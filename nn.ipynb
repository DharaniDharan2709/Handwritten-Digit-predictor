```python
import torch
from torchvision import datasets
from torchvision.transforms import ToTensor
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
from torch.utils.data import DataLoader




training_data = datasets.MNIST(
    root="data",
    train=True,
    download=True,
    transform=ToTensor()
)

test_data = datasets.MNIST(
    root="data",
    train=False,
    download=True,
    transform=ToTensor()
)


batch_size = 64

training_loader = DataLoader(
    training_data,
    batch_size=batch_size,
    shuffle=True
)

test_loader = DataLoader(
    test_data,
    batch_size=batch_size,
    shuffle=False
)
```

```python
from torch import nn
class DigitRecognizer(nn.Module):
    def __init__(self):
        super().__init__()

        self.flatten = nn.Flatten()
        self.hidden_layer = nn.Linear(784, 30)
        self.activation = nn.Sigmoid()
        self.output_layer = nn.Linear(30, 10)

    def forward(self, images):
        x = self.flatten(images)
        z1 = self.hidden_layer(x)
        a1 = self.activation(z1)
        logits = self.output_layer(a1)

        return logits


model = DigitRecognizer().to(device)

print(model)
```

```python
loss_function = nn.CrossEntropyLoss()

optimizer = torch.optim.SGD(
    model.parameters(),
    lr=0.1
)


```

```python

def train_one_epoch():
    model.train()

    total_loss = 0.0
    correct = 0
    total = 0

    for images, labels in training_loader:
        images = images.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()

        logits = model(images)
        loss = loss_function(logits, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()

        predictions = logits.argmax(dim=1)
        correct += (predictions == labels).sum().item()
        total += labels.size(0)

    average_loss = total_loss / len(training_loader)
    accuracy = correct / total

    return average_loss, accuracy
```

```python

def test_model():
    model.eval()

    total_loss = 0.0
    correct = 0
    total = 0

    with torch.no_grad():
        for images, labels in test_loader:
            images = images.to(device)
            labels = labels.to(device)

            logits = model(images)
            loss = loss_function(logits, labels)

            total_loss += loss.item()

            predictions = logits.argmax(dim=1)
            correct += (predictions == labels).sum().item()
            total += labels.size(0)

    average_loss = total_loss / len(test_loader)
    accuracy = correct / total

    return average_loss, accuracy
```

```python

number_of_epochs = 5

for epoch in range(number_of_epochs):
    training_loss, training_accuracy = train_one_epoch()
    test_loss, test_accuracy = test_model()

    print(f"Epoch {epoch + 1}/{number_of_epochs}")
    print(f"  Training loss:     {training_loss:.4f}")
    print(f"  Training accuracy: {training_accuracy:.2%}")
    print(f"  Test loss:         {test_loss:.4f}")
    print(f"  Test accuracy:     {test_accuracy:.2%}")


```

```python


# --------------------------------------------------
# 9. Save learned parameters
# --------------------------------------------------

torch.save(
    model.state_dict(),
    "digit_recognizer.pth"
)

print("Model saved.")
```
